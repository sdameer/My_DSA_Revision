```python
from collections import deque
```

`No of Provinces : `

```python
def NoOfProvinces(isConnected):
    v = len(isConnected)
    # convert the matrix into adj
    adj = [[] for _ in range(v)]
    for i in range(v):
        for j in range(v):
            if i != j and isConnected[i][j] == 1 :
                adj[i].append(j)


    ans = 0
    visited = [False] * v
    def dfs(node):
        visited[node] = True

        # go deep , loop through all of its connected neighbors
        for neighbor in adj[node]:
            if not visited[neighbor]:
                dfs(neighbor)


    for i in range(v):
        if not visited[i]:
            # we've got a new graph so start dfs
            ans += 1
            dfs(i)

    return ans

```

`No of Islands : `

```python
def NoOfIslands(grid) :
    count = 0 
    n = len(grid) # rows
    m = len(grid[0]) # cols
    visited = [[False] * m for _ in range(n)] # 2d matrix

    def bfs(row , col):
        visited[row][col] = True
        q = deque([(row , col)])

        while q :
            x , y = q.popleft()
            # explore 4 neighbors
            for dr , dc in [(-1 , 0), (0 , 1), (1 , 0), (0 , -1)]:
                nrow = x + dr
                ncol = y + dc
                inside_mat = (nrow >= 0 and nrow < n and ncol >= 0 and ncol < m)
                if inside_mat and not visited[nrow][ncol] and grid[nrow][ncol] == "1" :
                    visited[nrow][ncol] = True
                    q.append((nrow , ncol))


    for row in range(n):
        for col in range(m):
            if not visited[row][col] and grid[row][col] == "1":
                count += 1
                bfs(row , col)
    return count
```
