`No of Provinces : `

```python
def NoOfProvinces(isConnected):
    v = len(isConnected)
    # convert the matrix into adj
    adj = [[] for _ in range(v)]
    for i in range(v):
        for j in range(v):
            if i != j and isConnected[i][j] == 1 :
                adj[i].append(j)


    ans = 0
    visited = [False] * v
    def dfs(node):
        visited[node] = True

        # go deep , loop through all of its connected neighbors
        for neighbor in adj[node]:
            if not visited[neighbor]:
                dfs(neighbor)


    for i in range(v):
        if not visited[i]:
            # we've got a new graph so start dfs
            ans += 1
            dfs(i)

    return ans

```

`No of Islands : `

```python
def NoOfIslands(grid) :
    count = 0 
    n = len(grid) # rows
    m = len(grid[0]) # cols
    visited = [[False] * m for _ in range(n)] # 2d matrix

    def bfs(row , col):
        visited[row][col] = True
        q = deque([(row , col)])

        while q :
            x , y = q.popleft()
            # explore 4 neighbors
            for dr , dc in [(-1 , 0), (0 , 1), (1 , 0), (0 , -1)]:
                nrow = x + dr
                ncol = y + dc
                inside_mat = (nrow >= 0 and nrow < n and ncol >= 0 and ncol < m)
                if inside_mat and not visited[nrow][ncol] and grid[nrow][ncol] == "1" :
                    visited[nrow][ncol] = True
                    q.append((nrow , ncol))


    for row in range(n):
        for col in range(m):
            if not visited[row][col] and grid[row][col] == "1":
                count += 1
                bfs(row , col)
    return count
```

`Flood Fill :`

```python
def floodFill(image , sr, sc, color):
    nr = len(image)
    nc = len(image[0])
    image_copy = [row[:] for row in image]
    visited = [[False] * nc for _ in range(nr)]
    start_color = image[sr][sc]

    def dfs(row, col):
        visited[row][col] = True
        image_copy[row][col] = color

        for dr , dc in [(-1,0),(0,1),(1,0),(0,-1)]:
            nrow = row + dr
            ncol = col + dc
            inside_mat = (nrow >=0 and nrow <nr and ncol >= 0 and ncol <nc)
            if inside_mat and not visited[nrow][ncol] and image_copy[nrow][ncol] == start_color:                    
                dfs(nrow, ncol)
                


    dfs(sr , sc)
    return image_copy
```

`Rotten Oranges : `

```python
def orangesRotting(grid):
        n = len(grid)
        m = len(grid[0])
        q = deque()
        visited = [[0] * m for _ in range(n)]
        fresh = 0
        for i in range(n):
            for j in range(m):
                if grid[i][j] == 2 :
                    q.append((i , j , 0))
                    visited[i][j] = 2
                elif grid[i][j] == 1 :
                    fresh += 1 # we are keeping track of fresh oranges so that we can compare later if all oranges were rotten then we can return time if not -1

        tm = 0
        rotten = 0

        while q :
            x , y, t = q.popleft()
            tm = t

            for dr , dc in [(-1,0),(1,0),(0,1),(0,-1)]:
                nrow = x + dr
                ncol = y + dc
                inside_mat = (0<= nrow < n and 0<= ncol <m)
                if inside_mat and  visited[nrow][ncol] != 2 and grid[nrow][ncol] == 1 :
                    rotten +=1
                    visited[nrow][ncol] = 2
                    q.append((nrow , ncol , t+1))

        
        if rotten != fresh : return -1
        return tm
```

`Detect a Cycle : `

```python
def cycleDetection(edges, n, m):
    adj = [[] for _ in range(n + 1)]

    for u, v in edges:
        adj[u].append(v)
        adj[v].append(u)

    visited = [False] * (n + 1)

    def bfs(node):
        q = deque([(node, -1)])
        visited[node] = True

        while q:
            node, parent = q.popleft()

            for neighbor in adj[node]:
                if not visited[neighbor]:
                    visited[neighbor] = True
                    q.append((neighbor, node))
                elif neighbor != parent:
                    return True

        return False

    for i in range(1, n ):
        if not visited[i]:
            if bfs(i):
                return "Yes"

    return "No"
```

`0 1 Matrix : `

```python
def updateMatrix(mat): 
        n = len(mat)
        m = len(mat[0])
        q = deque()
        dist = [[0] * m for _ in range(n)]
        visited = [[False] * m for _ in range(n)]
        # loop through the matrix and as True in visited where ever we find an 0
        # and add the all the 0 nodes to the queue to process 
        for i in range(n):
            for j in range(m):
                if mat[i][j] == 0:
                    visited[i][j] = True
                    q.append((i , j , 0))

        # start bfs : 
        while q  : 
            x , y, t = q.popleft()
            # check all four sides 
            for dr ,dc in [(-1,0),(1, 0),(0,-1), (0,1)]:
                nrow = x + dr
                ncol = y + dc
                inside_matrix = (0<=nrow<n and 0<= ncol < m)
                if inside_matrix and not visited[nrow][ncol] and mat[nrow][ncol] == 1:
                    q.append((nrow, ncol, t+1))
                    visited[nrow][ncol] = True
                    dist[nrow][ncol] = t+1


        return dist
```

`No of Enclaves : `

```python
def numEnclaves(grid):
    n = len(grid)
    m = len(grid[0])
    count = 0
    visited = [[False]*m for _ in range(n)]

    
    def dfs(row , col):
        visited[row][col] = True

        for dr , dc in [(-1,0),(1,0),(0,-1),(0,1)]:
            nrow = row + dr
            ncol = col + dc
            inside_mat = (0<=nrow<n and 0<=ncol<m)
            if inside_mat and not visited[nrow][ncol] and grid[nrow][ncol] == 1 :
                dfs(nrow , ncol)


    # top row
    for i in range(m):
        if grid[0][i] == 1 and not visited[0][i] :
            dfs(0,i)
    # bottom row
    for i in range(m):
        if grid[n-1][i] == 1 and not visited[n-1][i]:
            dfs(n-1 ,i)
    # right column
    for i in range(n):
        if grid[i][0] == 1 and not visited[i][0] :
            dfs(i , 0)
    # left column
    for i in range(n):
        if grid[i][m-1] == 1 and not visited[i][m-1]:
            dfs(i,m-1)

    
    for row in range(n):
        for col in range(m):
            if not visited[row][col] and grid[row][col] == 1 :
                count += 1

    return count
```

`Distinct Islands`

```python
def distinctIsland(arr,n,m) :
    visited = [[False] * m for _ in range(n)]
    count = set()

    def dfs(row , col , base_row , base_col , shape):
        visited[row][col] = True

        shape.append((row - base_row , col - base_col ))

        for dr , dc in [(-1,0), (1,0), (0,-1),(0,1)]:
            nrow = dr + row
            ncol = dc + col
            inside_mat = (0<=nrow <n and 0<=ncol <m)
            if inside_mat and not visited[nrow][ncol] and arr[nrow][ncol] == 1:
                dfs(nrow , ncol, base_row , base_col , shape)

    for row in range(n):
        for col in range(m):
            if not visited[row][col] and arr[row][col] ==1 :
                shape = []
                # here base row and base col is start of the dfs when ever we find a new shape
                dfs(row ,col , row , col , shape)
                count.add(tuple(shape))

    return len(count)
```

`Bipartite Graph (BFS)`

```python
def isBipartite(self, graph):
    n = len(graph)
    visited = [-1] * n

    def bfs(start_node):
        visited[start_node] = 0
        q = deque([start_node])

        while q :
            node = q.popleft()
            for neighbor in graph[node]:
                if visited[neighbor] ==  -1:
                    visited[neighbor] = 1 - visited[node]
                    q.append(neighbor)
                elif visited[neighbor] ==  visited[node]:
                    return False
        return True

    for i in range(n):
        if visited[i] == -1:
            if not bfs(i):
                return False
        
    return True
```

`Bipartite Graph (DFS)`

```python
def isBipartite(self, graph):
    n = len(graph)
    visited = [-1] * n

    def dfs(node):
    
        for neighbor in graph[node]:
            if visited[neighbor] ==  -1:
                visited[neighbor] = 1 - visited[node]
                if not dfs(neighbor):
                    return False
            elif visited[neighbor] ==  visited[node]:
                return False
        return True

    for i in range(n):
        if visited[i] == -1:
            if not dfs(i):
                return False
        
    return True
```

`Detect Cycle in a Directed Graph using DFS`

```python
def detectCycleInDirectedGraph(n, edges):
    adj = [[] for _ in range(n+1)]
    visited = [False] * (n+1)
    path = [False] * (n+1)

    for u , v in edges :
        adj[u].append(v)
    
    def dfs(node):
        visited[node] = True
        path[node] = True

        for neighbor in adj[node]:            
            if not visited[neighbor]:
                if dfs(neighbor) :
                    return True
            elif visited[neighbor] and path[neighbor]:
                return True
        path[node] = False
        return False

        
            

    for i in range(n):
        if not visited[i]:
            if dfs(i):
                return True

    return False
```

`Find Eventual Safe States (DFS)`

```python
def eventualSafeNodes( graph):
    n = len(graph)
    path = [False] * n
    safe = [False] * n
    visited = [False] * n
    
    ans = []

    def dfs(node):
        visited[node] = True
        path[node] = True

        for neighbor in graph[node]:
            if not visited[neighbor]:
                if dfs(neighbor):
                    return True
            elif path[neighbor]:
                return True
        
        # when it is not a loop then save the SAFE nodes
        safe[node] = True
        # backtrack
        path[node] = False
        return False

    for i in range(n):
        if not visited[i]:
            dfs(i)
    
    for i in range(n):
        if safe[i]:
            ans.append(i)

    return ans
```

`Topological Sort  (DFS)  :`

```python
def topologicalSort(adj, v, e):
    # assume adj 
    n = len(adj)
    stack = []
    visited = [False] * n 

    def dfs(node):
        visited[node] = True

        for neighbor in adj[node]:
            if not visited[neighbor]:
                dfs(neighbor)
        stack.append(node)

    for i in range(n):
        if not visited[i]:
            dfs(i)

    return stack[::-1]
```

`Topological Sort  (BFS)  :`

```python
from collections import deque

def topoSort(V, adj):
    n = V
    indegree = [0] * n

    for neighbors in adj:
        for node in neighbors:
            indegree[node] += 1

    q = deque()

    for i in range(n):
        if indegree[i] == 0:
            q.append(i)

    topo = []

    while q:
        node = q.popleft()
        topo.append(node)

        for neighbor in adj[node]:
            indegree[neighbor] -= 1
            if indegree[neighbor] == 0:
                q.append(neighbor)

    return topo
```
