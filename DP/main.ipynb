````python
def factorial(n):
    if n ==0 or n==1 :
        return n
    return factorial(n-1) + factorial(n-2)
````

`Tabulation : `

````python
def fibonacci(n):
    if n == 0 or n == 1:
        return n

    dp = [0] * (n + 1)

    dp[0] = 0
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
````

`Tabulation O(1) :`

````python
def fibonacci(n):
    if n == 0:
        return 0

    prev2 = 0   # fib(0)
    prev = 1    # fib(1)

    for i in range(2, n + 1):
        curr = prev + prev2
        prev2 = prev
        prev = curr

    return prev
````

````python
def ways(n):
    if n == 0 :
        return 1
    
    dp = [0] * (n + 1) # create a table
    
    dp[0] = 1
    dp[1] = 1


    for i in range(1, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
            
    return dp[n]


ways(3)
````

`Tabulation O(1) :`

````python
def ways(n):
    if n == 0:
        return 1

    prev2 = 1  # ways(0)
    prev = 1   # ways(1)

    for i in range(2, n + 1):
        curr = prev2 + prev
        prev2 = prev
        prev = curr

    return prev

ways(5)
````

````python
# ----- Top Down approach -----
def func(idx , heights, dp):
    if idx == 0 :return 0

    # when there is a value return ; why recursion
    if dp[idx] != -1  : return dp[idx]

    left = func(idx-1 , heights, dp) + abs(heights[idx] - heights[idx-1])
    if (idx > 1) :
        right = func(idx - 2 , heights, dp) + abs(heights[idx] - heights[idx-2])
    else : 
        right = float('inf')
    # here we can get negative values so prevent it 

    dp[idx] = min(left, right)
    return dp[idx]
    
    
def frogJump(n: int, heights):
    dp = [-1] * (n+1)
    return func(n-1 , heights, dp)
````

`Tabulation : `

````python
def frogJump(n, heights):
    dp = [0] * n

    for i in range(1, n):
        left = dp[i - 1] + abs(heights[i] - heights[i - 1])

        right = float('inf')
        if i > 1:
            right = dp[i - 2] + abs(heights[i] - heights[i - 2])

        dp[i] = min(left, right)

    return dp[n - 1]
````

````python
def frogJump(n: int, heights):
    prev1 =0 
    prev2 = 0

    for i in range(1, n):
        left = prev1 + abs(heights[i] - heights[i - 1])

        right = float('inf')
        if i > 1:
            right = prev2 + abs(heights[i] - heights[i - 2])

        curr = min(left, right)
        prev2 = prev1
        prev1 = curr

    return prev1
````

`Unique Paths : `

````python
class Solution:
    def uniquePaths(self, m: int, n: int):
        def func(i, j):
            if i == 0 and j == 0:
                return 1
            if not (0 <= i < n and 0 <= j < m):
                return 0

            up = func(i-1, j)
            left = func(i , j-1)
            return left + up

        return func(n - 1, m - 1)
````

````python
class Solution:
    def uniquePaths(self, m: int, n: int):
        dp = [[-1]*m for _ in range(n)]

        def func(i, j):
            if i == 0 and j == 0:
                return 1
            if not (0 <= i < n and 0 <= j < m):
                return 0
            
            if dp[i][j] != -1 :
                return dp[i][j]

            up = func(i-1, j)
            left = func(i , j-1)
            dp[i][j] = left + up
            return dp[i][j]

        return func(n - 1, m - 1)
````

````python
class Solution:
    def uniquePaths(self, m: int, n: int):
        dp = [[-1]*m for _ in range(n)]        
        dp[0][0] = 1

        for i in range(n):
            for j in range(m):
                if i ==0 and j == 0 :
                    continue  
                          
                up = dp[i-1][j] if i > 0 else 0
                left =  dp[i][j-1] if j > 0 else 0
                dp[i][j] = left + up
        
        return dp[n-1][m-1]
    
````
