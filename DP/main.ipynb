## `Fibonacci : `

`Recursion :`

```python
def factorial(n):
    if n ==0 or n==1 :
        return n
    return factorial(n-1) + factorial(n-2)
```

`Memoization :`

```python

dp = {}
def fibonacci(n, dp={}):
    if n == 0 or n == 1:
        return n

    if n in dp:
        return dp[n]

    dp[n] = fibonacci(n - 1,dp) + fibonacci(n - 2,dp )
    return dp[n]


for i in range(10):
    print(fibonacci(i), end=" ")
```

`Tabulation : `

```python
def fibonacci(n):
    if n == 0 or n == 1:
        return n

    dp = [0] * (n + 1)

    dp[0] = 0
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
```

`Tabulation O(1) :`

```python
def fibonacci(n):
    if n == 0:
        return 0

    prev2 = 0   # fib(0)
    prev = 1    # fib(1)

    for i in range(2, n + 1):
        curr = prev + prev2
        prev2 = prev
        prev = curr

    return prev
```

```python
count = 0
def fun(add , target):
    
    global count
    if add == target : 
        count += 1
        return
    
    for i in range(1, 3):
        if add+i <= target:        
            fun(add+i , target)
           
fun(0 , 3) 
print(count)

# here ther are only two paths for this question
# either 1 or 2 
# we can write an equation like fibonacci
# if we are going to add up we would have 
# used for loop - also an way used here
# to write relatino we subtract
```

```python

def ways(n):
    if n ==0 : return 1 # we got answer
    if n < 0 : return 0 # we have exceeded answer
    
    return ways(n-1) + ways(n-2)

ways(3)
```

`Memoization :`

```python
def ways(n,dp={}):
    if n in dp :
        return dp[n]
    
    if n ==0 : return 1 # we got answer
    if n < 0 : return 0 # we have exceeded answer
    
    dp[n] = ways(n-1,dp) + ways(n-2,dp)
    return dp[n]

ways(3)
```

`Tabulation : `

```python
def ways(n):
    if n == 0 :
        return 1
    
    dp = [0] * (n + 1) # create a table
    
    dp[0] = 1
    dp[1] = 1


    for i in range(1, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
            
    return dp[n]


ways(3)
```

`Tabulation O(1) :`

```python
def ways(n):
    if n == 0:
        return 1

    prev2 = 1  # ways(0)
    prev = 1   # ways(1)

    for i in range(2, n + 1):
        curr = prev2 + prev
        prev2 = prev
        prev = curr

    return prev

ways(5)
```

# `Frog Jump :`

`Recursion :`

```python
def func(idx , heights):
    if idx == 0 :
        return 0

    left = func(idx-1 , heights) + abs(heights[idx] - heights[idx-1])    
    right = float('inf')
    # here we can get negative values so prevent it 
    if (idx > 1) :
        right = func(idx - 2 , heights) + abs(heights[idx] - heights[idx-2])

    return min(left, right)
    
    
def frogJump():
    heights = [10 , 20 , 30 , 10]
    n = len(heights)
    return func(n-1 , heights)

frogJump()
```

`Memoization :`

```python
# ----- Top Down approach -----
def func(idx , heights, dp):
    if idx == 0 :return 0

    # when there is a value return ; why recursion
    if dp[idx] != -1  : return dp[idx]

    left = func(idx-1 , heights, dp) + abs(heights[idx] - heights[idx-1])
    if (idx > 1) :
        right = func(idx - 2 , heights, dp) + abs(heights[idx] - heights[idx-2])
    else : 
        right = float('inf')
    # here we can get negative values so prevent it 

    dp[idx] = min(left, right)
    return dp[idx]
    
    
def frogJump(n: int, heights):
    dp = [-1] * (n+1)
    return func(n-1 , heights, dp)
```

`Tabulation : `

```python
def frogJump(n, heights):
    dp = [0] * n

    for i in range(1, n):
        left = dp[i - 1] + abs(heights[i] - heights[i - 1])

        right = float('inf')
        if i > 1:
            right = dp[i - 2] + abs(heights[i] - heights[i - 2])

        dp[i] = min(left, right)

    return dp[n - 1]
```

`Tabulation O(1) :`

```python
def frogJump(n: int, heights):
    prev1 =0 
    prev2 = 0

    for i in range(1, n):
        left = prev1 + abs(heights[i] - heights[i - 1])

        right = float('inf')
        if i > 1:
            right = prev2 + abs(heights[i] - heights[i - 2])

        curr = min(left, right)
        prev2 = prev1
        prev1 = curr

    return prev1
```
