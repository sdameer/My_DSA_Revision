`Recursion :`

```python
def factorial(n):
    if n ==0 or n==1 :
        return n
    return factorial(n-1) + factorial(n-2)
```

`Memoization :`

```python
def fibonacci(n, dp={}):
    if n == 0 or n == 1:
        return n

    if n in dp:
        return dp[n]

    dp[n] = fibonacci(n - 1, dp) + fibonacci(n - 2, dp)
    return dp[n]
```

`Tabulation : `

```python
def fibonacci(n):
    if n == 0 or n == 1:
        return n

    dp = [0] * (n + 1)

    dp[0] = 0
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
```

`Tabulation O(1) :`

```python
def fibonacci(n):
    if n == 0:
        return 0

    prev2 = 0   # fib(0)
    prev = 1    # fib(1)

    for i in range(2, n + 1):
        curr = prev + prev2
        prev2 = prev
        prev = curr

    return prev
```
