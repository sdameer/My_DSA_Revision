`Unique Paths : `

````python
class Solution:
    def uniquePaths(self, m: int, n: int):
        def func(i, j):
            if i == 0 and j == 0:
                return 1
            if not (0 <= i < n and 0 <= j < m):
                return 0

            up = func(i-1, j)
            left = func(i , j-1)
            return left + up

        return func(n - 1, m - 1)
````

````python
class Solution:
    def RodCutting(self, price, n):

        def func(idx , tar):
            # at idx = 0 rod len = 1 ; N (tar)
            # if after all calls we are left with an N value as we already know that the at idx = 0 ; rod = 1
            # so we need to use that rod len = 1 * n times 
            # so that the rod will cut into n pieces => n=4 =>[1,1,1,1] so at each we need price
            # so n * price[0]
            if idx == 0 :return tar*price[0]

            not_take = func(idx-1 , tar)
            rod_length = idx + 1
            take = 0 
            if rod_length <= tar:
                take = price[idx] + func(idx  , tar - rod_length)
            return max(take , not_take)

        n = len(price)
        return func(n-1 , n)
    
    # we can write the commom pattern Memoiation + Tabulation + Space opt
````
