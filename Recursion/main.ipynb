```python
def func(idx , arr , stack , curr_sum , target , n):
    if idx == n:
        if curr_sum == target :
            return 1 
        return 0
    # print(f"{idx} == {n} -> {stack}")

    stack.append(arr[idx])
    curr_sum += arr[idx]

    left = func(idx + 1 ,arr , stack , curr_sum , target , n)

    stack.pop()
    curr_sum -= arr[idx]

    right = func(idx + 1 , arr , stack , curr_sum , target , n)

    return left + right

arr = [1 , 2, 1]
n = len(arr)
stack = []
idx = 0
target = 2
curr_sum = 0
func(idx , arr , stack , curr_sum , target , n)
```

```python
def merge_sort(arr, low, high):
    if low >= high:
        return

    mid = (low + high) // 2

    merge_sort(arr, low, mid)
    merge_sort(arr, mid + 1, high)

    merge(arr, low, mid, high)


def merge(arr, low, mid, high):
    temp = []
    left, right = low, mid + 1

    while left <= mid and right <= high:
        if arr[left] <= arr[right]:
            temp.append(arr[left])
            left += 1
        else:
            temp.append(arr[right])
            right += 1

    while left <= mid:
        temp.append(arr[left])
        left += 1

    while right <= high:
        temp.append(arr[right])
        right += 1

    for i in range(low, high + 1):
        arr[i] = temp[i - low]


arr = [5, 2, 8, 1, 3]
merge_sort(arr, 0, len(arr) - 1)
print(arr)
```
