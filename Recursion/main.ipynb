```python
def print_subsequence(i, arr, path, curr_sum, k):
    if i == len(arr):
        if curr_sum == k:
            print(path)
        return

    # Take
    path.append(arr[i])
    print_subsequence(i + 1, arr, path, curr_sum + arr[i], k)
    path.pop()

    # Not Take
    print_subsequence(i + 1, arr, path, curr_sum, k)


arr = [1, 2, 1]
k = 2

print_subsequence(0, arr, [], 0, k)
```

```python
def one_subsequence(i, arr, path, curr_sum, k):
    if i == len(arr):
        if curr_sum == k:
            print(path)
            return True
        return False

    # Take
    path.append(arr[i])
    if one_subsequence(i + 1, arr, path, curr_sum + arr[i], k):
        return True
    path.pop()

    # Not Take
    if one_subsequence(i + 1, arr, path, curr_sum, k):
        return True

    return False


arr = [1, 2, 1]
k = 2

one_subsequence(0, arr, [], 0, k)
```

```python
def count_subsequence(i, arr, curr_sum, k):
    if i == len(arr):
        return 1 if curr_sum == k else 0

    # Take
    left = count_subsequence(i + 1, arr, curr_sum + arr[i], k)

    # Not Take
    right = count_subsequence(i + 1, arr, curr_sum, k)

    return left + right


arr = [1, 2, 1]
k = 2

print(count_subsequence(0, arr, 0, k))
```

```python
def merge_sort(arr, low, high):
    if low >= high:
        return

    mid = (low + high) // 2

    merge_sort(arr, low, mid)
    merge_sort(arr, mid + 1, high)

    merge(arr, low, mid, high)


def merge(arr, low, mid, high):
    temp = []
    left, right = low, mid + 1

    while left <= mid and right <= high:
        if arr[left] <= arr[right]:
            temp.append(arr[left])
            left += 1
        else:
            temp.append(arr[right])
            right += 1

    while left <= mid:
        temp.append(arr[left])
        left += 1

    while right <= high:
        temp.append(arr[right])
        right += 1

    for i in range(low, high + 1):
        arr[i] = temp[i - low]


arr = [5, 2, 8, 1, 3]
merge_sort(arr, 0, len(arr) - 1)
print(arr)
```
