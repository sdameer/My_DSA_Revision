```python
def merge_sort(arr, low, high):
    if low >= high:
        return

    mid = (low + high) // 2

    merge_sort(arr, low, mid)
    merge_sort(arr, mid + 1, high)

    merge(arr, low, mid, high)


def merge(arr, low, mid, high):
    temp = []
    left, right = low, mid + 1

    while left <= mid and right <= high:
        if arr[left] <= arr[right]:
            temp.append(arr[left])
            left += 1
        else:
            temp.append(arr[right])
            right += 1

    while left <= mid:
        temp.append(arr[left])
        left += 1

    while right <= high:
        temp.append(arr[right])
        right += 1

    for i in range(low, high + 1):
        arr[i] = temp[i - low]


arr = [5, 2, 8, 1, 3]
merge_sort(arr, 0, len(arr) - 1)
print(arr)
```

`Combination sum - I`

```python
candidates = [2 , 3]
target = 6
ans = []

def func(idx , target , stack):
    # -ve target is useless so go back
    if target < 0 : 
        return 
    if idx == len(candidates):
        # when len is equal to current idx :
        # you've came to an end so now stop
        if target == 0 :
            ans.append(stack[:]) # save a deep copy
        return 
    curr_num = candidates[idx]
    stack.append(curr_num)
    # take 
    func(idx , target-curr_num , stack )

    stack.pop()

    # don't take 
    func(idx+1 ,target , stack)

func(0 , target , [])
ans
```

`Combination sum - II`

```python
# In Combination Sum I, we often reach the answer after recursively deciding
# whether to take or skip every element, so the base case is usually:
# if idx == len(candidates)

# Here, each recursive call uses a for-loop to choose the next element.
# As soon as target becomes 0, we've already formed a valid combination.
# We don't need to wait until idx reaches the end.

# if there are duplicate numbers it will do the recursio twice on same two numbers
candidates = [2, 3, 4]
target = 6
candidates.sort()
ans = []


def func(idx, target, stack):
    if target < 0:
        return

    # if idx == len(candidates):
    if target == 0:
        ans.append(stack[:])
        return

    # take current num and explore all other values till end
    for i in range(idx, len(candidates)):
        # Skip duplicate numbers at the SAME recursion level.
        # The first occurrence (i == idx) is always allowed.
        if i != idx and candidates[i] == candidates[i-1]:
            continue

        # what is the point of finding a number in array which is greater than sum
        if candidates[i] > target:
            break

        curr_num = candidates[i]
        stack.append(curr_num)
        func(i + 1, target - curr_num, stack)
        stack.pop()


func(0, target, [])
ans
```

`Subset sum - II`

```python
nums = [1,2,3]
nums.sort()
ansList = []
    
def findSubsets( ind, nums, ds, ansList):
    ansList.append(ds[:])  # Make a copy of the current subset

    for i in range(ind, len(nums)):
        # Skip duplicates at the same recursion level
        if i != ind and nums[i] == nums[i - 1]:
            continue

        ds.append(nums[i])
        findSubsets(i + 1, nums, ds, ansList)
        ds.pop()  # Backtrack

findSubsets(0, nums, [], ansList)
ansList
```

```python
nums = [1,2,3]
ans = []
ds = []
freq = [False] * len(nums)

def backtrack():
    if len(ds) == len(nums):
        ans.append(ds[:])
        return 
    # we need permutations so go from start to end all the time 
    for i in range(len(nums)):
        # if means that we have not used it yet
        if freq[i] == False :
            freq[i] = True
            ds.append(nums[i])
            backtrack()

            ds.pop()
            freq[i] = False

backtrack()
ans
```

`N - Queen : `

```python

def can_i_place_queen(  row , col , board , n):
    # check the current row is there any Q ?
    c = col
    while c >= 0 :
        if board[row][c] == "Q":
            return False            
        c -= 1

    # check upper diagonal
    r  , c  = row , col 
    while r >= 0 and c >= 0:
        if board[r][c] == "Q":
            return False
        c -= 1
        r -= 1
    
    r , c = row , col 
    while r <n and c >= 0:
        if board[r][c] == "Q":
            return False
        c -= 1
        r += 1

    return True

def solveNQueens( n: int) :
    # we work on this board and when ever we find an solution we add it to answer 
    board = [["."]*n for _ in range(n)]        
    ans = []
    
    def solve(col):
        if col == n :
            ans.append(["".join(row) for row in board])
            return 
        
        # from left to right 
        # start row : 0 to 4 
        for row in range(n):
            if can_i_place_queen(row , col, board , n):
                board[row][col] = "Q"
                # now you have placed the queen in a col
                # with this placement lets move to another col
                # because this is backtracking 
                solve(col + 1)
                # agar fail hogayeh toh dusra col try karo
                # lekin usse pehle poorana queen toh nikal do
                board[row][col] = "."

    # let's start with col = 0
    solve(0)    
    return ans

ans
```

`Sudoku Solver : `

```python

def solve( board):
    # take a row 
    for row in range(9):
        # then go all over it columns L -> R
        for col in range(9):
            # if it is empty
            if board[row][col] == ".":
                # take one by one from 1 to 9 
                for num in "123456789":
                    # check whether the no is in row , col , 3x3 its square
                    if is_valid(board, row, col, num): 
                        # when it is not there then place it 
                        board[row][col] = num
                        # when the board is valid and even the solve return True it 
                        # means we have solved it 
                        # if we have solved the board then don't try with other possibilities
                        # when we pass board as param we are telling that 
                        # hey! , restart the work from 0,0 till you found an empty cell recursively
                        if solve(board):
                            return True

                        # this is simple because we are backtracking we try with other num s
                        # so we are removing it when a number fails 
                        board[row][col] = "."
                # I cannot solve the Sudoku starting from this empty cell with any digit.
                # this could occur when we need a number next but we have used it before 
                return False
    return True

def is_valid( board, row, col, num):


    for i in range(9):

        if board[i][col] == num:
            return False

        if board[row][i] == num:
            return False

        if board[3 * (row // 3) + i // 3][3 * (col // 3) + i % 3] == num:
            return False

    return True
```
